1\. Write a function that converts number representation, bin<->dec<->hex. (Clearly using the corresponding python built-in functions is not fair..)

```python
# prefisso esadecimali: 0x. Prefisso binari: 0b. Per i negativi basta il - davanti al prefisso.
#i bit sono infiniti: dopo il prefisso posso continuare all'infinito



def bin_to_dec(numero,n):
    dec=0
    for i,cifra in enumerate(reversed(numero)):   #CAPOVOLGO LA STRINGA. con enumerate, i mi dà l'indice e parte da 0
        dec += int(cifra) *(2**i)
    return dec
    
def dec_to_bin(numero):
    num=int(numero)
    if num == 0:
        return 0
    binario = ""
    while num > 0:
        resto = num % 2
        # Attacca il resto a sinistra della stringa
        binario = str(resto) + binario
        num = num // 2  # Divisione intera
    return binario

def dec_to_hex(numero):
    num=int(numero)
    if num == 0:
        return 0
    mappa = "0123456789abcdef"
    esadecimale = ""
    while num > 0:
        resto = num % 16
        esadecimale = mappa[resto] + esadecimale
        num = num // 16
    return esadecimale

    
def hex_to_dec(numero,n):
    mappa = "0123456789abcdef"
    dec = 0
    for i,carattere in enumerate(reversed(numero.lower())):   #con lower abbasso eventuali maiuscole
        valore_numerico= mappa.index(carattere)  #CONVERSIONE: cerca in "mappa" il carattere e assegna la posizione come valore
        dec += valore_numerico*(16**i)
    return dec
   
     

tipo = input("si vuole inserire un numero dec, bin o hex? ")

if tipo in ("bin", "dec", "hex"):
    numero = input("inserire il numero: ")   #è una stringa!!
    n=len(numero)
    if tipo == "bin":
        print(f"decimale: {bin_to_dec(numero,n)}")
        print(f"esadecimale: {dec_to_hex(bin_to_dec(numero,n))}")
    elif tipo == "dec":
        print(f"binario: {dec_to_bin(numero)}")
        print(f"esadecimale: {dec_to_hex(numero)}")
    elif tipo == "hex":
        print(f"decimale: {hex_to_dec(numero,n)}")
        print(f"binario: {dec_to_bin(hex_to_dec(numero,n))}")
else:
    print("carattere non valido. Interruzione del programma")
```

```python
word="000000000000000000000000000000000"

def conversion(word):
    #sign
    if int(word[0])==1:
        sign = -1
    else:
        sign =1
   
    #mantissa
    mantissa_string = word[9:32]
    fraction=0.0 #inizializzo la frazione
    for i, bit in enumerate(mantissa_string, start=1):   #enumerate chiede due variabili: una per l'indice (i) e una per il valore.
        if bit == '1':                                   #start=1 è opzionale, serve a partire l'indice da 1 (i=1) anzichè 0
              fraction += 2.0 ** (-i)     
    
    #exponent
    exponent_bits = word[1:9]
    E = int(exponent_bits, 2)   #interpreta la stringa come intero binario e la converte in decimale

                                     

    if E == 255:
        if fraction == 0:
            return float('-inf') if sign == -1 else float('inf')     #overflow
        else:
            return float('nan')                                      #operazione non valida
    elif E == 0:         # Numero subnormale / denormalizzato  --> sono quelli molto vicini allo zero!!
        return sign * (2.0**-126) * fraction
    else:
    # Numero normalizzato standard
        return sign * (2.0 ** (E - 127)) * (1.0 + fraction)
        
print(conversion(word))
```

3\. Write a program to determine the underflow and overflow limits (within a factor of 2) for python on your computer. 

**Tips**: define two variables inizialized to 1 and halve/double them enough time to exceed the under/over-flow limits  

```python
#FUORI ESERCIZIO

import sys
print(sys.float_info)
```

```python
# --- 1. OVERFLOW (raddoppia finché non esplode) ---
overflow = 1.0
while overflow != float("inf"):
    ultimo_overflow = overflow
    overflow = overflow * 2  # raddoppia

#raffinamento
overflow = ultimo_overflow
while overflow != float("inf"):
    ultimo_overflow = overflow
    overflow = overflow * 1.01  

print("Limite di overflow (valore massimo):", ultimo_overflow)
#ora molto simile a sys.float_info



# --- 2. UNDERFLOW (dimezza finché non si azzera) ---
underflow = 1.0
while underflow > 0.0:
    ultimo_underflow = underflow
    underflow = underflow / 2  # QUI DIVIDIAMO PER 2! 

#raffinamento
underflow = ultimo_underflow
while underflow > 0.0:
    ultimo_underflow = underflow
    underflow = underflow / 2

print("Limite di underflow (valore minimo prima di 0):", ultimo_underflow)
#diverso da sys.float_info perchè questo comprende anche i numeri subnormali, mentre l'altro si fermava a quelli normalizzati
```

4\. Write a program to determine the machine precision

```python
epsilon = 1.0

# Finché 1.0 + epsilon è diverso da 1.0, continuiamo a rimpicciolirlo
while 1.0 + epsilon != 1.0:
    ris = epsilon
    epsilon = epsilon / 2

print("La precisione di macchina è:", ris)
```

5\. Write a function that takes in input three parameters $a$, $b$ and $c$ and prints out the two solutions to the quadratic equation $ax^2+bx+c=0$ using the standard formula:
$$
x=\frac{-b\pm\sqrt{b^2-4ac}}{2a}
$$

(a) use the program to compute the solution for $a=0.001$, $b=1000$ and $c=0.001$

(b) re-express the standard solution formula by multiplying top and bottom by $-b\mp\sqrt{b^2-4ac}$ and again find the solution for $a=0.001$, $b=1000$ and $c=0.001$. How does it compare with what previously obtained? Why?

(c) write a function that compute the roots of a quadratic equation accurately in all cases

```python
import math

a,b,c=0.001,1000,0.001

def sol(a,b,c):
    delta=b**2 -4*a*c
    if delta<0:
        return "imp"
    else:
        x1=(-b+math.sqrt(delta))/(2*a)
        x2=(-b-math.sqrt(delta))/(2*a)
    return x1,x2

#in this example we have that -b+sqrt(delta) is near the zero. So

def solNew(a,b,c):
    delta=b**2 -4*a*c
    if delta<0:
        return "imp"
    else:
        x1=(2*c) / (-b-math.sqrt(delta))
        x2=(2*c) / (-b+math.sqrt(delta))
    return x1,x2

def solNewDef(a,b,c):
    delta=b**2 -4*a*c
    if delta<0:
        return "imp"
    else:
        if b>0 or b==0:
            x1=(2*c) / (-b-math.sqrt(delta))
            x2=(-b-math.sqrt(delta))/(2*a)
        else:
            x1=(-b+math.sqrt(delta))/(2*a)
            x2=(2*c) / (-b+math.sqrt(delta))
    return x1,x2
    
print(sol(a,b,c))
print(solNew(a,b,c))
print(solNewDef(a,b,c))
```

WHY?
because in this example, where sqrt(delta) is near 1000, the error between the real quantity sqrt(delta) and the quantity memorized as a double is very small, but it becomes relevant when we subtract b, which is exactly 1000. Therefore the error is amplified and the solution is not correct (x1) so we can use the other formula.
In the case b<0, we would have the opposite situation for x1,x2.

(a) Calculate the derivative of the function at the point $x = 1$ using the derivative definition:

$$
\frac{{\rm d}f}{{\rm d}x} = \lim_{\delta\to0} \frac{f(x+\delta)-f(x)}{\delta}
$$

with $\delta = 10^{−2}$. Calculate the true value of the same derivative analytically and compare with the answer your program gives. The two will not agree perfectly. Why not?

(b) Repeat the calculation for $\delta = 10^{−4}, 10^{−6}, 10^{−8}, 10^{−10}, 10^{−12}$ and $10^{−14}$. How does the accuracy scales with $\delta$?

```python
x=1.0

def f(x,delta):
    return x*(x-1)

derAnal = 2*x-1
print(f"analytical derivative: {derAnal} \n"
     f"function derivative:")

for i in range (-2,-16,-2):  #il terzo numero stabilisce i passi intermedi
    delta=10**(i)
    derivative = ( f(x + delta,delta) - f(x,delta) )/delta
    print("delta=10**",i,"->",derivative)


#for x=1 -> f'(1) = 1+delta  so error between the two results is equal to delta.
#as delta decrease, the accuracy becomes better and better until 10**-8.
#here the accuracy decreases until 10**-14, while for 10**-16 the function returns 0.0  (delta < machineepsilon=2.22*10**-16=precision of the machine).

#WHAT HAPPENS?
#as delta decreases, f(x+delta)-f(x) is rounded to a multiple of epsilon (machine precision). Thus, many digits are lost and the error is amplified
#by the small delta in the denominator.
```

7\. Consider the integral of the semicircle of radius 1:
$$
I=\int_{-1}^{1} \sqrt(1-x^2) {\rm d}x
$$
which it's known to be $I=\frac{\pi}{2}=1.57079632679...$.
Alternatively we can use the Riemann definition of the integral:
$$
I=\lim_{N\to\infty} \sum_{k=1}^{N} h y_k 
$$

with $h=2/N$ the width of each of the $N$ slices the domain is divided into and where
$y_k$ is the value of the function at the $k-$th slice.

(a) Write a programe to compute the integral with $N=100$. How does the result compares to the true value?

(b) How much can $N$ be increased if the computation needs to be run in less than a second? What is the gain in running it for 1 minute? 


```python
import math, time


def I(n):
    h=2/n
    somma=0
    for i in range(0,n+1):
        x = -1 + i*h
        somma += math.sqrt(1-x**2)   
    return somma*h

print(f"True value: {math.pi/2} \n"
    f"function integral:")

i,t=0,0
while t<1.0:
    if t<1.0:
        n=100*10**i
        t0 = time.perf_counter()
        integral=I(n)
        t = time.perf_counter() - t0 
        print("n=10**",2+i,"->",integral,"time:",t, "sec")
        i += 1

#il tempo è proporzionale a n
#L'errore di integrazione diminuisce proporzionalmente a h = 2/n, quindi aumentano le cifre decimali esatte.

#per un minuto basta moltiplicare per 60: n=10**6 * 60sec.
#si guadagna solo una cfra significativa in più rispetto a 7 sec

t0 = time.perf_counter()
integral=I(60*10**6)
t = time.perf_counter() - t0 
print("meno di un minuto:",integral,"time:",t)
```
