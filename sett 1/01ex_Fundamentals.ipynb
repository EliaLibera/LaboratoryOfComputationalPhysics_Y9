2\. Convert the following function into a pure function with no global variables or side effects

````python
#ATTENZIONE!!: una variabile globale è una variabile definita nel main, fuori dalla funzione, ovvero può essere usata da qualsiasi funzione.
#FUNZIONE PURA: no effetti collaterali (non cambia le variabili) e dipende solo dai suoi argomenti (non da variabili esterne)
````

````python
x = 5
def f(alist):
    for i in range(x):
         alist.append(i)
    return alist

alist = [1,2,3]
ans = f(alist)
print (ans)
print (alist) # alist has been changed!
````

````python
def f(alist, x):
    new_list = alist.copy() 
    
    for i in range(x):
        new_list.append(i)
        
    return new_list

alist = [1, 2, 3]
x = 5

ans = f(alist, x)

print("Risultato della funzione:", ans)   # Output: [1, 2, 3, 0, 1, 2, 3, 4]
print("Lista originale intatta:", alist) # Output: [1, 2, 3]
````

3\. Write a `decorator` hello that makes every wrapped function print “Hello!”, i.e. something like:

```python
@hello
def square(x):
    return x*x
```



````python
#DECORATOR: modificare una funzione esistente senza doverne riscrivere o toccare il codice interno.
#suo argomento: un'altra funzione -> aggiunge del comportamento extra -> restituisce una nuova funzione modificata.
#richiede una funzione interna (spesso chiamata wrapper) che esegue l'azione extra e poi chiama la funzione originale.

def hello(func):    #usa la funzione (a cui collegherò @hello) come argomento
    def wrapper(*args):    #*args cattura un numero qualsiasi di argomenti sottoforma di tupla
        print("Hello!")      # Comportamento aggiunto (stampa Hello!)
        return func(*args)       # Esegue la funzione originale passandole l'argomento
    return wrapper

#Applicazione del decoratore a qualche funzione
@hello
def square(x):
    return x * x
risultato = square(5)
print(risultato)

@hello
def printsomething():
    return "Stampa qualcosa"
stringa = printsomething()
print(stringa)

@hello
def somma(a,b):
    return a+b
risultatosomma=somma(2,4)
print(risultatosomma)




````

4\. Write the factorial function so that it a) does and b) does not use recursion.

````python
#CON RECURSION
def factorial(n):
    if n==0 or n==1:
        return 1
    else:
        return factorial(n-1)*n
        
print([factorial(i) for i in range(0,10)])   #ricordarsi LA LISTA!!!
````

````python
#SENZA RECURSION

def factorial(n):
    ris=1
    for i in range(1,n+1):  #da 1 a n-1 compresi
        ris=ris*i
    return ris

print([factorial(i) for i in range(0,10)])   #ricordarsi LA LISTA!!!
````


where the entries of the dictionary's values are the linear, superficial and volumetric densities of the materials respectively.

In particular define a list of three lambda functions using a comprehension that computes the circumference, the area and the volume for a given radius.


````python
#zip: iterare elementi matchati di liste diverse

import math
densities = {"Al":[0.5,1,2],"Fe":[3,4,5],"Pb": [15,20,30]}
#dizionario: .keys() restituisce solo le chiavi, .values() i valori, .items() restituisce sia la chiave che il valore (tuple) (chiave, valore).
radii = [1,2,3]

computation=[(math.pi*2,1),(math.pi,2),(math.pi*4/3,3)]             #lista (matrice) contenente (operazione,potenza)

hypervolume_functions=[lambda r, a=a, b=b: a*(r**b) for a, b in computation]  #LISTA DI FUNZIONI
#quando chiamo le 3 funzioni, a e b conservano il loro ultimo valore (quello del volume). 
#perciò mentre scorro le 3 funzioni devo far si che a,b abbiano di volta in volta il giusto valore 
#dunque scrivo a=a, b=b.
for r in radii:
    print("Raggio: ",r)
    for materiale,dens_list in zip(densities.keys(),densities.values()):   #match tra nome materiale e relativa lista di densità (keys and values)
        weights=[hypervolume(r)*d for hypervolume, d in zip(hypervolume_functions, dens_list)]  #match tra lin/sup/vol con il giusto ipervolume
        print(materiale,"->","Peso [cerchio,disco,volume] = ",weights)




````
