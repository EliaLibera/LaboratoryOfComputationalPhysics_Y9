2\. The swap function

Write a function that swap the values of two input variables x and y (whatever the type). Try to do that also without a temporary variable

```python
a=input("Inserire la variabile a:")
b=input("Inserire la variabile b:")
def swap(a,b):
    c=a
    a=b
    b=c
    return a,b
a,b=swap(a,b)
print("a = ",a)
print("b = ",b)
```

3\. Computing the distance

Write a function that calculates and returns the euclidean distance between two points *u* and *v*, where *u* and *v* are both 2-tuples *(x,y)*. For example, if *u=(3,0)* and *v=(0,4)*, the function should return 5

```python
import math
print("Inserire le coordinate del punto u")
x1=float(input("x:"))
y1=float(input("y:"))
print("Inserire le coordinate del punto v")
x2=float(input("x:"))
y2=float(input("y:"))
p=[x1,y1,x2,y2]
def distance(p):
    d=math.sqrt((p[0] - p[2])**2 + (p[1] - p[3])**2)
    return d
d=distance(p)
print("the distance between u and v is ",d)

#CONVERTIRE IN FLOAT L'INPUT!! infatti l'input è sempre stringa
```

4\. Counting letters

```python
s="Write a program that prints the numbers from 1 to 100. \
But for multiples of three print Mickey instead of the number and for the multiples of five print Mouse. \
For numbers which are multiples of both three and five print MickeyMouse"


s=s.lower()  #per avere tutte minuscole
caratteri_contati=[]
for carattere in s:
    if carattere not in caratteri_contati:
        caratteri_contati.append(carattere)    #aggiungere carattere alla LISTA
        print(carattere,": ",s.count(carattere)," volte") #contare carattere STRINGA
        
```

5\. Isolating the unique

Write a function that determines and count the unique numbers in the list *l*

```python
l = [36, 45, 58, 3, 74, 96, 64, 45, 31, 10, 24, 19, 33, 86, 99, 18, 63, 70, 85,
 85, 63, 47, 56, 42, 70, 84, 88, 55, 20, 54, 8, 56, 51, 79, 81, 57, 37, 91,
 1, 84, 84, 36, 66, 9, 89, 50, 42, 91, 50, 95, 90, 98, 39, 16, 82, 31, 92, 41,
 45, 30, 66, 70, 34, 85, 94, 5, 3, 36, 72, 91, 84, 34, 87, 75, 53, 51, 20, 89, 51, 20]

def unici(l):
    risultato = []
    for num in l:
        if l.count(num) == 1:  #conto se compare una volta!!
            risultato.append(num)
    return risultato

numeri_unici = unici(l)
print("I numeri unici sono:", numeri_unici)
print("e sono", len(numeri_unici), "numeri")

#NB: non rimuovere elementi da una lista mentre la sto scorrendo!!
```

6\. Combination of functions

Write two functions - one that returns the square of a number, and one that returns the cube. Now write a third function that returns the number raised to the 6th power using the two previous functions.

```python
a=float(input("inserire un numero:"))
def square(a):
    return a**2
def cube(a):
    return a**3
def potenzaSesta(a):
    return(cube(square(a)))
#usare come argomento di una funzione il return di un'altra funzione!!
print("il suo quadrato è",square(a))
print("il suo cubo è",cube(a))
print("elevato alla sesta dà",potenzaSesta(a))
```

8\. Nested list comprehension

A Pythagorean triple is an integer solution to the Pythagorean theorem $a^2+b^2=c^2$. The first Pythagorean triple is (3,4,5). Find and put in a tuple all unique Pythagorean triples for the positive integers a, b and c less than 100.

```python
triple = tuple((a, b, c)      
               for a in range(1, 100)
               for b in range(a + 1, 100)
               for c in range(b + 1, 100)
               if a**2 + b**2 == c**2)

print(triple)
print("Ci sono", len(triple), "terne")
#le tonde sono un generatore (come le list, ma con le tonde).  Tuple "consuma" il generatore creando la tupla.
#si capisce che è nested list perchè ci sono un for dentro l'altro.
```

9\. Normalization

Write a function that takes a tuple of numbers and returns it with the entries normalized to one

```python
a= (4,5,6)
def normalizzazione(a):
    somma=sum(a)
    return tuple( x / somma for x in a)
print(normalizzazione(a))
```
